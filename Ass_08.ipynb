```python

# Discount factor
gamma = 0.9

# Policy
policy = {
    "S0": "up",
    "S1": "up",
    "S2": "right",
    "S3": "right",
    "S4": "right"
}

# Environment
def step(state, action):
    transitions = {
        ("S0", "up"): ("S1", 5),
        ("S1", "up"): ("S2", 4),
        ("S2", "right"): ("S3", 3),
        ("S3", "right"): ("S4", 2),
        ("S4", "right"): ("S5", 10)
    }
    return transitions[(state, action)]


def state_value_function(state):
    total = 0
    discount = 1
    visited = set()

    while state != "S5":
        if state in visited:
            print("Loop detected:", state)
            return None

        visited.add(state)

        action = policy[state]
        state, reward = step(state, action)

        total += discount * reward
        discount *= gamma

    return total


def action_value_function(state, action):
    next_state, reward = step(state, action)

    total = reward
    discount = gamma
    state = next_state
    visited = set()

    while state != "S5":
        if state in visited:
            print("Loop detected:", state)
            return None

        visited.add(state)

        action = policy[state]
        state, reward = step(state, action)

        total += discount * reward
        discount *= gamma

    return total


print("State Value of S0:", state_value_function("S0"))
print("Action Value of S0, Up:", action_value_function("S0", "up"))
```
