```python
import math
# Part A - Machine Initialization
machines = ['A', 'B', 'C']
# Number of plays
N_play = 10
# Initialize tables
N = {'A': 0, 'B': 0, 'C': 0}       # Number of times machine is played
R = {'A': 0, 'B': 0, 'C': 0}       # Total reward
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0} # Estimated reward
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}
print("UCB Multi-Armed Bandit")
print("Machines:", machines)
print("Total Plays:", N_play)
# Part B - Initial test
# Play each machine once so that UCB can be calculated
print("\n---Initial Test ---")
for machine in machines:
    reward = float(input(f"Play with machine {machine}\nEnter reward: "))
    N[machine] = 1
    R[machine] = reward
    Q[machine] = reward
# Part C, D and E - UCB Selection, Reward Collection and Q Update
for play in range(4, N_play + 1):
    print(f"\n Play {play}")
    # Calculate UCB values
    for machine in machines:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(play)) / N[machine]
        )
    print("\nCurrent UCB values:")
    for machine in machines:
        print(f"Machine {machine}: {UCB[machine]:.4f}")
    # Select machine having maximum UCB
    selected_machine = max(machines, key=lambda x: UCB[x])
    print(f"\nSelected Machine: {selected_machine}")
    # Part D - Get reward from user
    reward = float(input("Enter reward: "))
    # Part E - Update N, R and Q
    N[selected_machine] += 1
    R[selected_machine] += reward
    # Incremental sample-average formula
    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]
# Part F - Display Final Results
print("\n\nFINAL RESULTS")
print("\nMachine\tN\tR\tUCB\t\tQ")
for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]:.2f}\t"
        f"{UCB[machine]:.4f}\t\t"
        f"{Q[machine]:.4f}"
    )
# Total reward received
total_reward = sum(R.values())
print("\nTotal Reward Received:", total_reward)
print("\nAnalysis:")
for machine in machines:
    print(
        f"Machine {machine}: Played {N[machine]} times, "
        f"Total Reward = {R[machine]:.2f}, "
        f"Estimated Q-value = {Q[machine]:.4f}"
    )
```
