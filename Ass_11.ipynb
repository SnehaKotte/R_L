```python
gamma = 0.9
theta = 0.000001
states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']
policy = {
    'S0': 'Up',
    'S1': 'Up'
}
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}
def get_next_state(state, action):

    if state == 'S0':
        if action == 'Right':
            return 'S1'
        else:
            return 'S0'

    elif state == 'S1':
        if action == 'Right':
            return 'G'
        else:
            return 'S1'
def get_reward(next_state):

    if next_state == 'G':
        return 100
    else:
        return -1
iteration = 0

while True:

    iteration += 1

    print("\nPolicy Evaluation - Iteration", iteration)
    while True:

        delta = 0

        for state in states:

            old_value = V[state]

            action = policy[state]

            next_state = get_next_state(state, action)

            reward = get_reward(next_state)

            V[state] = reward + gamma * V[next_state]

            delta = max(delta, abs(old_value - V[state]))

        print("S0 =", round(V['S0'], 4),
              "S1 =", round(V['S1'], 4))

        if delta < theta:
            break
    policy_stable = True

    print("\nPolicy Improvement")

    for state in states:

        old_action = policy[state]

        q_values = {}

        for action in actions:

            next_state = get_next_state(state, action)

            reward = get_reward(next_state)

            q_values[action] = reward + gamma * V[next_state]

        best_action = max(q_values, key=q_values.get)

        policy[state] = best_action

        print(state, "Q values:", q_values)

        if old_action != best_action:
            policy_stable = False

    print("Updated Policy:", policy)

    if policy_stable:
        break
print("\nFinal Optimal State Values")

for state in states:
    print(state, "=", round(V[state], 4))

print("G = 0 (Terminal State)")

print("\nFinal Optimal Policy")

for state in states:
    print(state, "=", policy[state])
print("\nFinal Action Value Table")

for state in states:

    print("\n", state)

    for action in actions:

        next_state = get_next_state(state, action)

        reward = get_reward(next_state)

        q_value = reward + gamma * V[next_state]

        print(action, "=", round(q_value, 4))

print("\nBellman Optimality Verification")

for state in states:

    q_values = []

    for action in actions:

        next_state = get_next_state(state, action)

        reward = get_reward(next_state)

        q_value = reward + gamma * V[next_state]

        q_values.append(q_value)

    print(state, ": V(s) =", round(V[state], 4),
          "Max Q(s,a) =", round(max(q_values), 4))
```
