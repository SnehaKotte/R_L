```python
gamma = 0.9
theta = 0.000001
states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}
def get_next_state(state, action):

    if state == 'S0':

        if action == 'Down':
            return 'S1'
        else:
            return 'S0'

    elif state == 'S1':

        if action == 'Right':
            return 'G'
        elif action == 'Up':
            return 'S0'
        else:
            return 'S1'
def get_reward(next_state):

    if next_state == 'G':
        return 100
    else:
        return -1
iteration = 0
while True:
    iteration = iteration + 1
    old_values = V.copy()
    delta = 0

    print("\nIteration:", iteration)

    for state in states:

        q_values = []

        for action in actions:

            next_state = get_next_state(state, action)

            reward = get_reward(next_state)

            q = reward + gamma * old_values[next_state]

            q_values.append(q)

        V[state] = max(q_values)

        delta = max(delta, abs(V[state] - old_values[state]))

    print("S0 =", round(V['S0'], 4))
    print("S1 =", round(V['S1'], 4))

    if delta < theta:
        break
print("\nFinal Optimal State Values")

for state in states:
    print(state, "=", round(V[state], 4))

print("G = 0 (Terminal State)")
print("\nFinal Action Value Table")
policy = {}
for state in states:
    print("\nState:", state)
    best_value = float('-inf')
    best_action = ''
    for action in actions:
        next_state = get_next_state(state, action)
        reward = get_reward(next_state)
        q = reward + gamma * V[next_state]
        print(action, "->", next_state, "=", round(q, 4))
        if q > best_value:
            best_value = q
            best_action = action
    policy[state] = best_action
print("\nOptimal Policy")
for state in states:
    print(state, "=", policy[state])
print("\nBellman Optimality Verification")
for state in states:
    q_values = []
    for action in actions:
        next_state = get_next_state(state, action)
        reward = get_reward(next_state)
        q = reward + gamma * V[next_state]
        q_values.append(q)
    print(state, "V(s) =", round(V[state], 4),
          "Max Q(s,a) =", round(max(q_values), 4))
```
