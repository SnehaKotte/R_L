```python
import random
# --------------------------------------------------
# Part A - Machine Initialization
# --------------------------------------------------
machines = ['A', 'B', 'C']
M = 3
N_play = 10
epsilon = 0.2
# N = Number of times each machine is selected
N = {'A': 0, 'B': 0, 'C': 0}
# R = Total reward received from each machine
R = {'A': 0, 'B': 0, 'C': 0}
# Q = Estimated average reward of each machine
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
# Store total reward received
total_reward = 0
# --------------------------------------------------
# Part B, C, D and E
# --------------------------------------------------
print("==============================================")
print(" Modified Epsilon-Greedy Multi-Armed Bandit")
print("==============================================")
for play in range(1, N_play + 1):
    print("\n----------------------------------------------")
    print("Play", play)
    print("----------------------------------------------")
    # Generate random number between 0 and 1
    r = random.random()
    print("Random number r =", round(r, 4))
    # Find machine with highest Q-value
    max_q = max(Q.values())
    best_machines = [
        machine for machine in machines
        if Q[machine] == max_q
    ]
    highest_machine = random.choice(best_machines)
    # --------------------------------------------------
    # Modified Epsilon-Greedy Decision
    # --------------------------------------------------
    if r < epsilon:
        # Exploration:
        # Select randomly, but exclude highest-Q machine
        exploration_machines = [
            machine for machine in machines
            if machine != highest_machine
        ]
        selected_machine = random.choice(exploration_machines)
        print("Decision: EXPLORATION")
        print("Highest Q machine:", highest_machine)
    else:
        # Exploitation:
        # Select highest Q-value machine
        selected_machine = highest_machine
        print("Decision: EXPLOITATION")
    print("Selected Machine:", selected_machine)
    # --------------------------------------------------
    # Part D - Reward Collection
    # --------------------------------------------------
    reward = float(
        input("Enter reward obtained from Machine "
              + selected_machine + ": ")
    )
    # --------------------------------------------------
    # Update N table
    # --------------------------------------------------
    N[selected_machine] += 1
    # --------------------------------------------------
    # Update R table
    # --------------------------------------------------
    R[selected_machine] += reward
    # --------------------------------------------------
    # Part E - Incremental Sample-Average Formula
    #
    # Q(a) = Q(a) + [R - Q(a)] / N(a)
    # --------------------------------------------------
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )
    total_reward += reward
    print("\nAfter this play:")
    print("N =", N)
    print("R =", R)
    print("Q =", {
        machine: round(Q[machine], 2)
        for machine in machines
    })
# --------------------------------------------------
# Part F - Display Final Results
# --------------------------------------------------
print("\n\n==============================================")
print("             FINAL RESULTS")
print("==============================================")
print("\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Reward)")
print("--------------------------------------------------------------")
for machine in machines:
    print(
        machine,
        "\t",
        N[machine],
        "\t\t",
        R[machine],
        "\t\t",
        round(Q[machine], 2)
    )
print("--------------------------------------------------------------")
print("\nTotal Reward Received =", total_reward)
average_reward = total_reward / N_play
print("Average Reward per Play =", round(average_reward, 2))
print("\nFinal Best Machine:")
best_q = max(Q.values())
best_machines = [
    machine for machine in machines
    if Q[machine] == best_q
]
print("Machine", best_machines, "with Q-value =", round(best_q, 2))
```
